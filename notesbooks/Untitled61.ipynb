### 5. Intégration des DataFrames et Ingénierie de Caractéristiques

Nous allons maintenant préparer les dataframes `events_df`, `gkg_df` et `events_excel_df` pour les fusionner avec `daily_score_df`. L'objectif est d'agréger les informations par date afin d'enrichir notre base de données principale (`daily_score_df`) avec des caractéristiques pertinentes pour la prédiction.

#### 5.1 Préparation des colonnes de date

Nous nous assurons que toutes les colonnes de date sont au format `datetime` et qu'elles ne contiennent que la date (sans l'heure) pour une agrégation journalière cohérente.

```python
# S'assurer que les colonnes de date sont au format datetime et tronquées à la date seule
daily_score_df['SQLDATE'] = daily_score_df['SQLDATE'].dt.normalize()
events_df['SQLDATE'] = events_df['SQLDATE'].dt.normalize()
gkg_df['DATE'] = gkg_df['DATE'].dt.normalize()
events_excel_df['SQLDATE'] = events_excel_df['SQLDATE'].dt.normalize()

# Renommer la colonne 'DATE' de gkg_df en 'SQLDATE' pour une cohérence lors de la fusion
gkg_df.rename(columns={'DATE': 'SQLDATE'}, inplace=True)

print("Colonnes de date normalisées et renommées avec succès.")

# Afficher les informations pour vérifier les types de données et les premières lignes
print("\nInformations sur daily_score_df:")
daily_score_df.info()
print("\nInformations sur events_df:")
events_df.info()
print("\nInformations sur gkg_df:")
gkg_df.info()
print("\nInformations sur events_excel_df:")
events_excel_df.info()
```

#### 5.2 Agrégation de `events_df` par jour

Nous allons agréger les caractéristiques clés de `events_df` par `SQLDATE` pour obtenir des statistiques journalières.

```python
events_daily_agg = events_df.groupby('SQLDATE').agg(
    events_goldstein_mean=('GoldsteinScale', 'mean'),
    events_goldstein_sum=('GoldsteinScale', 'sum'),
    events_avgtone_mean=('AvgTone', 'mean'),
    events_avgtone_sum=('AvgTone', 'sum'),
    events_nummentions_sum=('NumMentions', 'sum'),
    events_numsources_sum=('NumSources', 'sum'),
    events_numarticles_sum=('NumArticles', 'sum'),
    events_isrootevent_sum=('IsRootEvent', 'sum'),
    events_count=('SQLDATE', 'count') # Nombre total d'événements par jour
).reset_index()

print("Agrégation journalière de events_df effectuée avec succès.")
print("Premières 5 lignes de events_daily_agg:")
display(events_daily_agg.head())
```

#### 5.3 Agrégation de `gkg_df` par jour

Nous allons agréger les caractéristiques clés de `gkg_df` par `SQLDATE` pour obtenir des statistiques journalières.

```python
gkg_daily_agg = gkg_df.groupby('SQLDATE').agg(
    gkg_tone_mean=('tone', 'mean'),
    gkg_tone_sum=('tone', 'sum'),
    gkg_pos_score_mean=('positive_score', 'mean'),
    gkg_neg_score_mean=('negative_score', 'mean'),
    gkg_polarity_mean=('polarity', 'mean'),
    gkg_themes_count_sum=('themes_count', 'sum'),
    gkg_has_crisis_theme_sum=('has_crisis_theme', 'sum'),
    gkg_count=('SQLDATE', 'count') # Nombre total d'entrées GKG par jour
).reset_index()

print("Agrégation journalière de gkg_df effectuée avec succès.")
print("Premières 5 lignes de gkg_daily_agg:")
display(gkg_daily_agg.head())
```

#### 5.4 Agrégation de `events_excel_df` par jour

Nous allons agréger les caractéristiques clés de `events_excel_df` par `SQLDATE` pour obtenir des statistiques journalières.

```python
events_excel_daily_agg = events_excel_df.groupby('SQLDATE').agg(
    excel_goldstein_mean=('GoldsteinScale', 'mean'),
    excel_goldstein_sum=('GoldsteinScale', 'sum'),
    excel_avgtone_mean=('AvgTone', 'mean'),
    excel_avgtone_sum=('AvgTone', 'sum'),
    excel_nummentions_sum=('NumMentions', 'sum'),
    excel_numsources_sum=('NumSources', 'sum'),
    excel_numarticles_sum=('NumArticles', 'sum'),
    excel_score_risque_mean=('Score risque', 'mean'),
    excel_score_risque_sum=('Score risque', 'sum'),
    excel_count=('SQLDATE', 'count') # Nombre total d'événements Excel par jour
).reset_index()

print("Agrégation journalière de events_excel_df effectuée avec succès.")
print("Premières 5 lignes de events_excel_daily_agg:")
display(events_excel_daily_agg.head())
```

#### 5.5 Fusion des DataFrames agrégés avec `daily_score_df`

Nous allons maintenant fusionner tous les dataframes agrégés avec `daily_score_df` en utilisant la colonne `SQLDATE` comme clé. Nous utiliserons un `left join` pour conserver toutes les dates de `daily_score_df` et gérerons les valeurs manquantes résultantes de la fusion.

```python
merged_df = daily_score_df.copy()

# Fusionner avec events_daily_agg
merged_df = pd.merge(merged_df, events_daily_agg, on='SQLDATE', how='left')

# Fusionner avec gkg_daily_agg
merged_df = pd.merge(merged_df, gkg_daily_agg, on='SQLDATE', how='left')

# Fusionner avec events_excel_daily_agg
merged_df = pd.merge(merged_df, events_excel_daily_agg, on='SQLDATE', how='left')

# Remplacer les NaN introduits par les fusions (lorsqu'il n'y avait pas d'événements pour une date donnée) par 0
# Pour les colonnes d'agrégation numérique, c'est généralement un choix sûr
merged_df.fillna(0, inplace=True)

print("Tous les dataframes ont été fusionnés avec succès.")
print("Informations sur le dataframe fusionné (merged_df):")
merged_df.info()
print("\nPremières 5 lignes du dataframe fusionné:")
display(merged_df.head())
```

### 6. Ingénierie de Caractéristiques Supplémentaires (Time-Based Features)

Pour la modélisation de séries temporelles, il est souvent bénéfique d'ajouter des caractéristiques basées sur le temps. Nous allons extraire des informations de la colonne `SQLDATE`.

```python
merged_df['year'] = merged_df['SQLDATE'].dt.year
merged_df['month'] = merged_df['SQLDATE'].dt.month
merged_df['day'] = merged_df['SQLDATE'].dt.day
merged_df['day_of_week'] = merged_df['SQLDATE'].dt.dayofweek # Lundi=0, Dimanche=6
merged_df['day_of_year'] = merged_df['SQLDATE'].dt.dayofyear
merged_df['week_of_year'] = merged_df['SQLDATE'].dt.isocalendar().week.astype(int)
merged_df['quarter'] = merged_df['SQLDATE'].dt.quarter
merged_df['is_weekend'] = merged_df['day_of_week'].isin([5, 6]).astype(int) # 1 si week-end, 0 sinon

print("Caractéristiques temporelles ajoutées avec succès.")
print("Premières 5 lignes du dataframe fusionné avec les nouvelles caractéristiques:")
display(merged_df.head())
```

### 7. Préparation des Données pour la Modélisation

Nous allons maintenant définir nos variables cibles (`y`) et nos caractéristiques (`X`), puis diviser le jeu de données en ensembles d'entraînement et de test de manière chronologique.

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_squared_error, accuracy_score, classification_report

# Convertir 'status' en une variable cible numérique pour la détection de crise
# Par exemple: 0 pour 'Tension', 1 pour 'Crise'
merged_df['status_encoded'] = merged_df['status'].map({'Tension': 0, 'Crise': 1})
print("Encodage de la colonne 'status' effectué.")
print(merged_df['status_encoded'].value_counts())

# Supprimer les lignes où 'status_encoded' est NaN pour la classification
# Cette étape est cruciale car les modèles de scikit-learn ne gèrent pas les NaN dans la variable cible.
merged_df.dropna(subset=['status_encoded'], inplace=True)
print(f"\nNombre de lignes après suppression des NaN dans 'status_encoded': {len(merged_df)}")

# Définir la variable cible pour la prédiction temporelle (score)
y_score = merged_df['score']

# Définir la variable cible pour la détection de crise (status encodé)
y_status = merged_df['status_encoded']

# Définir les caractéristiques (X) en excluant les cibles et les colonnes non pertinentes
# 'SQLDATE' est exclue car nous avons extrait des caractéristiques temporelles
# 'status' est exclue car nous utilisons 'status_encoded' comme cible de classification
# 'score' et 'stability_score' sont exclues car ce sont nos cibles de régression/prédiction
features_to_drop = ['SQLDATE', 'status', 'score', 'stability_score', 'status_encoded']
X = merged_df.drop(columns=[col for col in features_to_drop if col in merged_df.columns])

# Gérer les colonnes non numériques qui pourraient être encore présentes dans X
# Pour cet exemple, nous allons simplement les supprimer. Dans un cas réel, on pourrait les encoder.
for col in X.columns:
    if X[col].dtype == 'object':
        X = X.drop(columns=[col])

print("\nVariables cibles et caractéristiques définies.")
print(f"Forme de X: {X.shape}")
print(f"Forme de y_score: {y_score.shape}")
print(f"Forme de y_status: {y_status.shape}")

# Split chronologique des données
# Utilisons 80% des données pour l'entraînement et 20% pour le test
train_size = int(len(merged_df) * 0.8)
X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_score_train, y_score_test = y_score.iloc[:train_size], y_score.iloc[train_size:]
y_status_train, y_status_test = y_status.iloc[:train_size], y_status.iloc[train_size:]

print(f"\nDonnées divisées en ensembles d'entraînement et de test (split temporel):")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_score_train shape: {y_score_train.shape}")
print(f"y_score_test shape: {y_score_test.shape}")
print(f"y_status_train shape: {y_status_train.shape}")
print(f"y_status_test shape: {y_status_test.shape}")
```

### 8. Construction et Évaluation des Modèles

Nous allons construire un modèle de régression pour prédire le `score` et un modèle de classification pour détecter le `status` de crise.

#### 8.1 Modèle de Prédiction du Score (Régression)

```python
# Initialiser et entraîner un modèle de régression (RandomForestRegressor)
model_score = RandomForestRegressor(n_estimators=100, random_state=42)
model_score.fit(X_train, y_score_train)

# Faire des prédictions sur l'ensemble de test
y_score_pred = model_score.predict(X_test)

# Évaluer le modèle
mse = mean_squared_error(y_score_test, y_score_pred)
rmse = np.sqrt(mse)

print(f"\nPerformance du modèle de prédiction du score:")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")

# Afficher les premières prédictions vs. valeurs réelles
print("\nComparaison des valeurs réelles et prédites (score):")
comparison_score = pd.DataFrame({'Réel': y_score_test, 'Prédit': y_score_pred})
display(comparison_score.head())
```

#### 8.2 Modèle de Détection de Crise (Classification)

```python
# Initialiser et entraîner un modèle de classification (RandomForestClassifier)
model_status = RandomForestClassifier(n_estimators=100, random_state=42)
model_status.fit(X_train, y_status_train)

# Faire des prédictions sur l'ensemble de test
y_status_pred = model_status.predict(X_test)

# Évaluer le modèle
accuracy = accuracy_score(y_status_test, y_status_pred)
classification_rep = classification_report(y_status_test, y_status_pred)

print(f"\nPerformance du modèle de détection de crise:")
print(f"Accuracy: {accuracy:.2f}")
print("\nRapport de classification:\n")
print(classification_rep)

# Afficher les premières prédictions vs. valeurs réelles
print("\nComparaison des valeurs réelles et prédites (statut):")
comparison_status = pd.DataFrame({'Réel': y_status_test, 'Prédit': y_status_pred})
display(comparison_status.head())
```
